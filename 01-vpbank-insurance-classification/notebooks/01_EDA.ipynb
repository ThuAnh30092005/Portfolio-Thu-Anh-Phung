## Categorical feature

```python
cat_feature = data[cat_feature + ["label"]]
cat_feature.head()

def plot_feature(df, col_name, target):
    f, ax2 = plt.subplots(nrows=1, ncols=1, figsize=(18, 5), dpi=90)

    # Use the original training_data DataFrame for grouping
    df_n = df.groupby([target, col_name]).size().reset_index()
    df_n = df_n.rename(columns={0: "count"})
    df_n["percentage"] = df_n['count'] / df_n.groupby(target)['count'].transform('sum') * 100

    sns.barplot(data=df_n, y="percentage", x=col_name, hue=target, ax=ax2)
    ax2.set_xlabel(col_name)
    ax2.set_ylabel('Distribution')
    plt.xticks(rotation=90)
    plt.tight_layout()

# Iterate through categorical features and plot
for i in cat_feature:  # Iterate through the list of column names
    if i != "label":
        # Use training_data as the DataFrame for plotting
        plot_feature(data, i, "label")
```
